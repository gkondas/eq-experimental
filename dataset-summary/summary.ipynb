```python
import pandas as pd
```

```python
subject_counts = pd.DataFrame(
    {
        "dataset": ["MIMIC", "Columbia", "NWICU"],
        "train": [200_773, 1_651_030, 20_723],
        "val": [25_059, 291_538, 2_590],
        "test": [15_059, 27_019, 772],
    }
)
subject_counts["total"] = subject_counts["train"] + subject_counts["val"] + subject_counts["test"]
subject_counts = subject_counts[["dataset", "train", "val", "test", "total"]]

subject_counts
```

```python
print(
    subject_counts.rename(
        columns={
            "dataset": "Dataset",
            "train": "Train",
            "val": "Val",
            "test": "Test",
            "total": "Total",
        }
    ).to_latex(
        index=False,
        formatters={c: "{:,}".format for c in ["Train", "Val", "Test", "Total"]},
        caption="Unique subject counts per split. Columbia reflects the intermediate (meicar / eq) preprocessing stage.",
        label="tab:dataset-summary",
        position="t",
    )
)
```

## LaTeX export

Tables are exported with pandas' built-in `DataFrame.to_latex`. Integer columns are formatted with thousands separators via `formatters`.

## Subject counts

Unique `subject_id` counts per split for each dataset. Columbia counts reflect the intermediate (meicar / eq) preprocessing stage.

## Events per split

Total event counts per split for each dataset.

```python
event_counts = pd.DataFrame(
    {
        "dataset": ["MIMIC", "Columbia", "NWICU"],
        "train": [625_289_578, 1_879_711_937, 43_992_928],
        "val": [77_489_324, 328_686_197, 5_573_365],
        "test": [76_954_727, 41_894_734, 5_219_071],
    }
)
event_counts["total"] = event_counts["train"] + event_counts["val"] + event_counts["test"]
event_counts = event_counts[["dataset", "train", "val", "test", "total"]]

event_counts
```

```python
print(
    event_counts.rename(
        columns={
            "dataset": "Dataset",
            "train": "Train",
            "val": "Val",
            "test": "Test",
            "total": "Total",
        }
    ).to_latex(
        index=False,
        formatters={c: "{:,}".format for c in ["Train", "Val", "Test", "Total"]},
        caption="Total event counts per split.",
        label="tab:events-per-split",
        position="t",
    )
)
```

## Vocabulary size

Vocabulary size for each dataset.

```python
vocab_sizes = pd.DataFrame(
    {
        "dataset": ["MIMIC", "Columbia", "NWICU"],
        "vocab_size": [11_476, 21_289, 2_159],
    }
)

vocab_sizes
```

```python
print(
    vocab_sizes.rename(
        columns={
            "dataset": "Dataset",
            "vocab_size": "Vocab",
        }
    ).to_latex(
        index=False,
        formatters={"Vocab": "{:,}".format},
        caption="Vocabulary size per dataset.",
        label="tab:vocab-size",
        position="t",
    )
)
```
